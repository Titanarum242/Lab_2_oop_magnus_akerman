# translate

```python
class Translate:

    def __init__(self, *numbers: float|int):
        for number in numbers:
            if not isinstance(number, (int, float)):
                raise TypeError(f"Expected int or float, got {type(number).__name__}")
        
        
        if len(numbers) > 3:
            raise ValueError("At most three numbers are allowed")

        
        self._numbers=tuple(float(number) for number in numbers)

        if len(self._numbers) ==0:
            self._numbers = (0.0, 0.0, 0.0)
        if len(self._numbers) ==1:
            self._numbers = (self._numbers[0],0.0, 0.0)
        if len(self._numbers) ==2:
            self._numbers = (self._numbers[0], self._numbers[1], 0.0)

    @property
    def x(self) -> float:
        return self._numbers[0] 
    @property
    def y(self) -> float:
        return self._numbers[1]
    @property
    def z(self) -> float:
        return self._numbers[2]

    def __eq__(self, other):
        if not isinstance(other, Translate):
            return NotImplemented
        return self._numbers == other._numbers
    
    def __getitem__(self, index: int) -> float:
        return self._numbers[index]

    def __repr__(self) -> str:
        return f"(x={self._numbers[0]}, y={self._numbers[1]}, z={self._numbers[2]})"
t1=Translate(1, 2, 3)
print(t1.x)
```

# circle

```python


import numpy as np
#from space_3d import Translate

class Circle:
    def __init__(self, radius: float, Translate: Translate) -> tuple:
        self._radius = radius
        self._translate = Translate

        if not isinstance(self._radius, (int, float)):
                raise TypeError(f"Expected int or float, got {type(self._radius).__name__}")
        if self._radius <= 0:
                raise ValueError("Radius must be a positive number")

    @property
    def area(self) -> float:
        return np.pi * self._radius ** 2

    @property
    def circumference(self) -> float:
        return 2 * np.pi * self._radius
    @property
    def circle_center(self) -> Translate:
        return self._translate
    @property
    def check_unity(self) -> bool:
        return self._radius == 1.0
        
    def __eq__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented
        return self._radius == other._radius and self._translate == other._translate

    def __repr__(self) -> str:
        return f"Circle(radius={self._radius}, center_coordinates:{self._translate})"

c1=Circle(99.0, Translate(-3,1))
c2=Circle(22.0, Translate(-22,88))
c3=Circle(55.0, Translate(-11,1))
c4=Circle(88.0, Translate(-17,1))

import matplotlib.pyplot as plt


circles = [c1, c2, c3, c4]

colors = ['r', 'b', 'y', 'g']

fig, ax = plt.subplots()

for circle, color in zip(circles, colors):


    patch = plt.Circle((circle.circle_center[0], circle.circle_center[1]), circle._radius, color=color, fill=False)

    ax.plot(circle.circle_center[0], circle.circle_center[1], 'o', color='black')
    ax.add_patch(patch)

    max_radius = max(c._radius for c in circles)

    ax.set_xlim(-10 - max_radius, 10 + max_radius)

    ax.set_ylim(-10 - max_radius, 10 + max_radius)

    ax.set_aspect('equal')

plt.show()

```

# rectangle

```python

import numpy as np
#from space_3d import Translate

class Rectangle:

    def __init__(self, base: float, height: float, Translate: Translate) -> tuple:
        self._base = base
        self._height = height
        self._translate = Translate

        if not isinstance(self._base and self._height, (int, float)):
            raise TypeError(f"Expected int or float, got {type(self._base).__name__}")
        if self._base <= 0 or self._height <= 0:
            raise ValueError("Base and height must be positive numbers")
        
        
    @property
    def area(self) -> float:
        return self._base * self._height

    @property
    def perimeter(self) -> float:
        return 2 * (self._base + self._height)

    @property
    def rectangle_center(self) -> Translate:
        center_x = self._translate.x + self._base / 2
        center_y = self._translate.y + self._height / 2
        return Translate(center_x, center_y)

    @property
    def check_square(self) -> bool:
        return self._base == self._height

    @property
    def check_unity(self) -> bool:
        return self._base == 1.0 and self._height == 1.0

    def __eq__(self, other):
        if not isinstance(other, Rectangle):
            return NotImplemented
        return self._height == other._height and self._base == other._base and self._translate == other._translate

    def __repr__(self) -> str:
        return f"Rectangle(base={self._base}, height={self._height}, center_coordinates:{self.rectangle_center})"

re1 = Rectangle(1, 2, Translate(1, 1))
print(re1)





```

# sphere

```python
import numpy as np

#from Lab_2_libs.cube import Cube
#from space_3d import Translate
class Sphere:
    
    def __init__(self, radius: float, Translate: Translate) -> tuple:
        self._radius = radius
        self._translate = Translate

        if not isinstance(self._radius, (int, float)):
            raise TypeError(f"Expected int or float, got {type(self._radius).__name__}")
        if self._radius <= 0:
            raise ValueError("Radius must be a positive number")

    @property
    def volume(self) -> float:
        return (4/3) * np.pi * self._radius**3

    @property
    def surface_area(self) -> float:
        return 4 * np.pi * self._radius**2
    @property
    def sphere_center(self) -> Translate:
        return self._translate
    @property
    def check_unity(self) -> bool:
        return self._radius == 1.0
    
    

    def __eq__(self, other):
        if not isinstance(other, Sphere):
            return NotImplemented
        return self._radius == other._radius and self._translate == other._translate

    def __repr__(self) -> str:
        return f"Sphere(radius={self._radius}, center_coordinates:{self._translate})"

# skall göras till en egen klass senare för sphere:
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

sp1 = Sphere(5, Translate(-11, 1, 8))
sp2 = Sphere(10, Translate(20, 5, 0))
sp3 = Sphere(7, Translate(0, 0, 0))

class Plot3D:

    def __init__(self):
        self._objects = []

    def add_sphere(self, sphere, color="r"):

        if not isinstance(sphere, Sphere):
            raise TypeError(
                f"Expected Sphere, got {type(sphere).__name__}"
            )

        self._objects.append((sphere, color))

    def plot(self):

        fig = plt.figure(figsize=(10, 8))
        ax = fig.add_subplot(111, projection="3d")

        for sphere, color in self._objects:

            x0 = sphere._translate.x
            y0 = sphere._translate.y
            z0 = sphere._translate.z
            r = sphere._radius

            u = np.linspace(0, 2 * np.pi, 40)
            v = np.linspace(0, np.pi, 40)

            x = x0 + r * np.outer(np.cos(u), np.sin(v))
            y = y0 + r * np.outer(np.sin(u), np.sin(v))
            z = z0 + r * np.outer(np.ones_like(u), np.cos(v))

            ax.plot_surface(x, y, z, color=color, alpha=0.3,linewidth=0)

            ax.scatter(x0, y0, z0, color="black", s=30)
           
        
        max_radius = max(sphere._radius for sphere, _ in self._objects)
        ax.set_xlim(-10 - max_radius, 10 + max_radius)
        ax.set_ylim(-10 - max_radius, 10 + max_radius)
        ax.set_zlim(-10 - max_radius, 10 + max_radius)
        ax.set_xlabel("X")
        ax.set_ylabel("Y")
        ax.set_zlabel("Z")

        ax.set_box_aspect([1, 1, 1])

        plt.show()




ph = Plot3D()

ph.add_sphere(sp1, "r")
ph.add_sphere(sp2, "g")
ph.add_sphere(sp3, "b")

ph.plot()
```

# cube

```python
class Cube:
    
    def __init__(self, side_length: float, Translate: Translate) -> tuple:
        self._side_length = side_length
        self._translate = Translate

        if not isinstance(self._side_length, (int, float)):
            raise TypeError(f"Expected int or float, got {type(self._side_length).__name__}")
        if self._side_length <= 0:
            raise ValueError("Side length must be a positive number")

    @property
    def volume(self) -> float:
        return self._side_length ** 3

    @property
    def surface_area(self) -> float:
        return 6 * self._side_length ** 2
    
    @property
    def cube_center(self) -> Translate:
        center_x = self._translate.x + self._side_length / 2
        center_y = self._translate.y + self._side_length / 2
        center_z = self._translate.z + self._side_length / 2
        return Translate(center_x, center_y, center_z)

    @property
    def check_unity(self) -> bool:
        return self._side_length == 1.0

    def __eq__(self, other):
        if not isinstance(other, Cube):
            return NotImplemented
        return self._side_length == other._side_length and self._translate == other._translate

    def __repr__(self) -> str:
        return f"Cube(side_length={self._side_length}, center_coordinates:{self.cube_center})"

cu1=Cube(2.0, Translate())
print(cu1) 


```
